Write a Python program to implement a simple calculator using user input and conditional statements.

```python
num1=int(input('enter first number'))
num2=int(input('enter secoend number'))
opreter=input('+,-,%,*')
if opreter=='+':
    print('sum of numbers',num1+num2)
    print(type(num1+num2))
elif opreter=='-':
    print('minus of numbers',num1-num2)
    print(type(num1-num2))
elif opreter=='/':
    if num2!=0:
        print('division of numbers',num1/num2)
        print(type(num1/num2))
elif opreter=='*':
    print('multiplication of number',num1*num2)
    print(type(num1*num2))
else:
    print('non')
```

Write a Python program that accepts marks of Mathematics, Science, English, Economics, and Hindi from the user and displays:

Total marks obtained
Percentage scored
Average marks
Grade based on percentage

```python
Mathematics=float(input('enter the Mathematics number'))
Science=float(input('enter the Science number'))
English=float(input('enter the English'))
Economics=float(input('enter the Economics number'))
Hindi=float(input('enter the Hindi number'))
total_marks=Mathematics+Science+English+Economics+Hindi
Percentage=(total_marks/500)*100
Average_marks=(total_marks)/5
print('Total marks obtained',total_marks)
print('Percentage scored',Percentage)
print('Average marks',Average_marks)
if Percentage >= 80:
    print('Grade based on percentage', Percentage, 'A')
elif Percentage >= 65:
    print('Grade based on percentage', Percentage, 'B')
elif Percentage >= 33:
    print('Grade based on percentage', Percentage, 'C')
else:
    print('Grade based on percentage', Percentage, 'Fail')
```

Discount Calculator(using Condition Loop) Question Write a program that asks the user for the total purchase amount. Apply discounts based on the following:

= 1000: 20% discount = 500: 10% discount = 100: 5% discount Less than 100: No discount Calculate and print the final amount after discount.

```python
purchase_amount=float(input('enter purchase_amount '))

if purchase_amount>=1000:
    print('purchase_amount with 20% discount',purchase_amount-(purchase_amount*0.20))

elif purchase_amount>=500 and purchase_amount<1000:
    print('purchase_amount with 10% discount',purchase_amount-(purchase_amount*0.10))

elif purchase_amount>=100:
    print('purchase_amount with 5% discount',purchase_amount-(purchase_amount*0.05))

else:
    print('no discount',purchase_amount)
```
